```python
import requests
import pandas as pd
from tqdm.notebook import tqdm

API_URL = "http://127.0.0.1:8000/api/search"
TOP_K = 20

MODEL_MAPPING = {
    "TF-IDF": "tfidf",
    "Word2Vec": "w2v",
    "Doc2Vec": "doc2vec",
    "LaBSE": "labse", 
    "BGE-M3": "bge",
    "Paraphrase": "mpnet",
    "Ensemble": "ensemble",
    "Word2Vec_sg": "w2v_sg",
    "Doc2Vec_dbow": "doc2vec_dbow",
}

TYPE_TO_SEARCH_MODE = {
    "basic": "title",    
    "upgrade": "overall" 
}

MODELS_CONFIG = [
    {"name": "TF-IDF",      "type": "basic"},   {"name": "TF-IDF",      "type": "upgrade"},
    {"name": "Word2Vec",    "type": "basic"},   {"name": "Word2Vec",    "type": "upgrade"},
    {"name": "Word2Vec_sg",    "type": "basic"},   {"name": "Word2Vec_sg",    "type": "upgrade"},
    {"name": "Doc2Vec",     "type": "basic"},   {"name": "Doc2Vec",     "type": "upgrade"},
    {"name": "Doc2Vec_dbow",     "type": "basic"},   {"name": "Doc2Vec_dbow",     "type": "upgrade"},
    {"name": "LaBSE",       "type": "basic"},   {"name": "LaBSE",       "type": "upgrade"},
    {"name": "BGE-M3",      "type": "basic"},   {"name": "BGE-M3",      "type": "upgrade"},
    {"name": "Paraphrase",  "type": "basic"},   {"name": "Paraphrase",  "type": "upgrade"},
    {"name": "Ensemble",    "type": "basic"},   {"name": "Ensemble",    "type": "upgrade"},
]

results_log = []


for config in MODELS_CONFIG:
    raw_name = config['name']  
    m_type = config['type']    
    
    # Lấy mã định danh API 
    model_short = MODEL_MAPPING.get(raw_name, raw_name.lower())
    
    # Xác định search_type
    search_mode = TYPE_TO_SEARCH_MODE.get(m_type, "overall")
    
    display_name = f"{model_short} ({search_mode})"
    
    # --- CHẠY TẬP TEST ---
    for q_idx, test_case in enumerate(tqdm(complex_test_set, desc=display_name, leave=False)):
        query_text = test_case['query']
        
        # Payload gửi đi
        payload = {
            "query": query_text,
            "model_name": model_short,  
            "search_type": search_mode,    
            "filters": {}
        }
        
        try:
            response = requests.post(API_URL, json=payload)
            
            if response.status_code == 200:
                predictions = response.json()
                
                # --- XỬ LÝ KẾT QUẢ ---
                if isinstance(predictions, list):
                    for rank, item in enumerate(predictions):
                        if rank >= TOP_K: break
                        
                        # Chỉ lấy các trường cần thiết
                        results_log.append({
                            "model_short": model_short, # Tên model ngắn (w2v, ensemble...)
                            "model_type": m_type,       # basic/upgrade
                            "query_id": q_idx,          # ID câu query
                            "query_text": query_text,   # Nội dung câu query
                            "rank": rank + 1,           # Thứ hạng
                            "job_id": item.get('id'),   # ID job
                            "score": item.get('similarity_score', item.get('score', 0)), 
                            "label": ""                 
                        })
            else:
                if q_idx == 0: 
                    print(f"❌ Backend lỗi {display_name}: {response.status_code}")
                
        except Exception as e:
            print(f"⚠️ Lỗi kết nối: {e}")

print("🎉 DONE! Đã thu thập xong.")

# --- LƯU FILE  ---
OUTPUT_FILE = "thucong.xlsx"
df_results = pd.DataFrame(results_log)

if not df_results.empty:
    final_columns = ["model_short", "model_type", "query_id", "query_text", "rank", "job_id", "score", "label"]
    df_results = df_results[final_columns]
    
    df_results.to_excel(OUTPUT_FILE)
    print(f" Đã lưu file: {OUTPUT_FILE}")
    print(" Hãy mở file này bằng Excel để bắt đầu gán nhãn cột 'label'.")
    display(df_results.head())
else:
    print("Không thu thập được dữ liệu nào.")
```

```python
# --- XÂY DỰNG GROUND TRUTH ---
from tqdm import tqdm
ground_truth_map = {} 
ground_truth_stats = [] # Để lưu thống kê xem query nào tìm được bao nhiêu job

print(f"Bắt đầu tạo Ground Truth cho {len(complex_test_set_2)} queries...")

for q_idx, test_case in enumerate(tqdm(complex_test_set_2)):
    # Lấy thông tin yêu cầu
    keywords = [k.lower() for k in test_case['must_have']]
    min_match = test_case['min_match']
    
    # --- LOGIC CỐT LÕI: ĐẾM TỪ KHÓA ---
    # Hàm này sẽ áp dụng cho từng dòng của DataFrame
    def is_relevant(text):
        if not isinstance(text, str): return False
        # Đếm số từ khóa xuất hiện trong văn bản
        match_count = sum(1 for kw in keywords if kw in text)
        return match_count >= min_match

    # Lọc ra các dòng thỏa mãn điều kiện
    relevant_jobs = df_gt[df_gt['full_text_search'].apply(is_relevant)]
    
    # Lấy tập hợp các ID
    relevant_ids = set(relevant_jobs['id'].tolist())
    
    # Lưu vào map
    ground_truth_map[q_idx] = relevant_ids
    
    # Lưu thống kê
    ground_truth_stats.append({
        "query_id": q_idx,
        "query_text": test_case['query'],
        "relevant_count": len(relevant_ids),
        "keywords": str(keywords)
    })

# --- BƯỚC 3: KIỂM TRA KẾT QUẢ ---
print("\n✅ HOÀN TẤT XÂY DỰNG GROUND TRUTH!")
```

```python
import pandas as pd
import numpy as np

def calculate_metrics_manual(group):
    # Lấy thông tin index
    if isinstance(group.name, tuple):
        model_name, q_id = group.name
    else:
        q_id = group.name
    
    # 1. Sắp xếp theo rank (1 -> 10)
    group = group.sort_values('rank')
    
    # Lấy nhãn (chỉ lấy tối đa 10 dòng đầu vì đánh giá thủ công Top 10)
    relevants = group['label'].values[:10]
    
    metrics = {}

    # ====================================================
    # A. MRR (Mean Reciprocal Rank)
    # ====================================================
    try:
        first_rel_idx = np.argmax(relevants == 1)
        if relevants[first_rel_idx] == 1:
            metrics['MRR'] = 1 / (first_rel_idx + 1)
        else:
            metrics['MRR'] = 0
    except:
        metrics['MRR'] = 0

    # ====================================================
    # B. METRICS TẠI CÁC NGƯỠNG K (5, 10)
    # ====================================================
    cutoffs = [5, 10]
    
    # Bảng discount cho NDCG
    global_discounts = np.log2(np.arange(len(relevants)) + 2)
    
    for k in cutoffs:
        actual_k = min(k, len(relevants))
        
        # Cắt mảng kết quả
        r_k = relevants[:actual_k]
        d_k = global_discounts[:actual_k]
        
        # --- 1. Hit Ratio@K (MỚI BỔ SUNG) ---
        # Ý nghĩa: Có ít nhất 1 kết quả đúng trong top K không?
        metrics[f'Hit@{k}'] = 1 if np.sum(r_k) > 0 else 0
        
        # --- 2. Precision@K ---
        metrics[f'P@{k}'] = np.sum(r_k) / k 
        
        # --- 3. NDCG@K ---
        dcg = np.sum(r_k / d_k)
        
        # IDCG (Lý tưởng: đưa hết các kết quả đúng tìm được lên đầu)
        ideal_r_k = sorted(r_k, reverse=True)
        idcg = np.sum(ideal_r_k / d_k)
        
        metrics[f'NDCG@{k}'] = dcg / idcg if idcg > 0 else 0

    return pd.Series(metrics)

# --- EXAMPLE USAGE ---
# manual_eval_df = pd.read_excel('manual_labels.xlsx') 
# report = manual_eval_df.groupby(['model', 'query_id']).apply(calculate_metrics_manual).reset_index()
# print(report.groupby('model').mean())
# --- HƯỚNG DẪN SỬ DỤNG ---
# Giả sử dataframe 'manual_eval_df' có các cột: [model, query_id, rank, label]
# rank: 1 đến 10
# label: 0 hoặc 1 (do người chấm)

final_report = dev_manual_short.groupby(['model', 'query_id']).apply(calculate_metrics_manual).reset_index()
summary = final_report.groupby('model').mean()
print(summary)
```

```python
import pandas as pd
import numpy as np

def calculate_metrics_optimized(group):
    # Lấy thông tin index
    if isinstance(group.name, tuple):
        model_name, q_id = group.name
    else:
        q_id = group.name
    
    # 1. Sắp xếp và lấy label
    group = group.sort_values('rank')
    relevants = group['label'].values 
    
    # 2. Lấy mẫu số (Ground Truth)
    # LƯU Ý QUAN TRỌNG: ground_truth phải là dict chứa set các ID đúng
    true_set = ground_truth.get(q_id, set())
    total_true_items = len(true_set)
    
    metrics = {}
    
    # --- Pre-calculation ---
    ranks = np.arange(len(relevants)) + 1
    cumulative_relevant = np.cumsum(relevants)
    precisions_at_i = cumulative_relevant / ranks
    discounts = np.log2(ranks + 1)
    
    # ====================================================
    # A. GLOBAL METRICS
    # ====================================================
    # MAP: Đúng (Đã chia cho total_true_items)
    if total_true_items > 0:
        metrics['MAP'] = np.sum(precisions_at_i * relevants) / total_true_items
    else:
        metrics['MAP'] = 0

    # MRR: Đúng (Không phụ thuộc số lượng true items, chỉ cần vị trí đầu tiên)
    try:
        first_rel_idx = np.argmax(relevants == 1)
        if relevants[first_rel_idx] == 1:
            metrics['MRR'] = 1 / (first_rel_idx + 1)
        else:
            metrics['MRR'] = 0
    except:
        metrics['MRR'] = 0

    # ====================================================
    # B. METRICS TẠI CÁC NGƯỠNG K
    # ====================================================
    cutoffs = [5, 10, 20]
    
    for k in cutoffs:
        actual_k = min(k, len(relevants))
        r_k = relevants[:actual_k]
        
        # 1. Hit Ratio@K (Đúng)
        metrics[f'Hit@{k}'] = 1 if np.any(r_k) else 0
        
        # 2. Precision@K (Đúng - chia cho K)
        metrics[f'P@{k}'] = cumulative_relevant[actual_k - 1] / k if actual_k > 0 else 0
        
        # 3. NDCG@K (ĐÃ SỬA LẠI LOGIC IDCG)
        # ---------------------------------------------------------
        dcg = np.sum(r_k / discounts[:actual_k])
        
        # Logic IDCG chuẩn:
        # Tạo một vector lý tưởng có độ dài bằng actual_k
        # Điền số 1 vào vector này dựa trên số lượng ground truth có sẵn
        ideal_r_k = np.zeros(actual_k)
        
        # Số lượng item đúng tối đa có thể nhét vào Top K này
        # Là min của (K đang xét, Tổng số item đúng thực tế)
        num_ones_possible = min(actual_k, total_true_items)
        
        ideal_r_k[:num_ones_possible] = 1 # Gán các vị trí đầu tiên bằng 1
        
        idcg = np.sum(ideal_r_k / discounts[:actual_k])
        
        metrics[f'NDCG@{k}'] = dcg / idcg if idcg > 0 else 0
        # ---------------------------------------------------------
        
        # 4. MRR@K (Đúng)
        first_idx_k = np.argmax(r_k == 1)
        if r_k[first_idx_k] == 1:
            metrics[f'MRR@{k}'] = 1 / (first_idx_k + 1)
        else:
            metrics[f'MRR@{k}'] = 0
            
        # 5. MAP@K (Đúng - Đã chia cho total_true_items)
        if total_true_items > 0:
            p_k = precisions_at_i[:actual_k]
            metrics[f'MAP@{k}'] = np.sum(p_k * r_k) / total_true_items
        else:
            metrics[f'MAP@{k}'] = 0

    return pd.Series(metrics)
```

```python
import pandas as pd
import numpy as np

def calculate_metrics_optimized(group):
    # Lấy thông tin index
    if isinstance(group.name, tuple):
        model_name, q_id = group.name
    else:
        q_id = group.name
    
    # 1. Sắp xếp và lấy label
    # Giả sử group này chứa tối đa N dòng (ví dụ N=20)
    group = group.sort_values('rank')
    relevants = group['label'].values 
    
    # 2. Lấy mẫu số (Ground Truth)
    true_set = ground_truth.get(q_id, set())
    total_true_items = len(true_set)
    
    metrics = {}
    
    # ====================================================
    # TÍNH TOÁN CÁC MẢNG CƠ SỞ (Pre-calculation)
    # Bước này tính 1 lần dùng cho cả Global và các Cutoff K
    # ====================================================
    
    # Mảng Rank: [1, 2, 3, ..., N]
    ranks = np.arange(len(relevants)) + 1
    
    # Mảng Precision tại mọi vị trí: [P@1, P@2, ..., P@N]
    # cumsum: số lượng đúng tích lũy
    cumulative_relevant = np.cumsum(relevants)
    precisions_at_i = cumulative_relevant / ranks
    
    # Mảng Discount cho NDCG: log2(i+2)
    discounts = np.log2(ranks + 1)
    
    # ====================================================
    # A. GLOBAL METRICS (Tương đương @MAX_LENGTH)
    # ====================================================
    if total_true_items > 0:
        # MAP = Tổng(Precision[i] * rel[i]) / Tổng số văn bản đúng
        metrics['MAP'] = np.sum(precisions_at_i * relevants) / total_true_items
    else:
        metrics['MAP'] = 0

    # MRR Global
    try:
        first_rel_idx = np.argmax(relevants == 1)
        if relevants[first_rel_idx] == 1:
            metrics['MRR'] = 1 / (first_rel_idx + 1)
        else:
            metrics['MRR'] = 0
    except:
        metrics['MRR'] = 0

    # ====================================================
    # B. METRICS TẠI CÁC NGƯỠNG K (CUTOFFS)
    # ====================================================
    cutoffs = [5, 10, 20]
    
    for k in cutoffs:
        # Nếu k lớn hơn độ dài list thì lấy max độ dài list
        actual_k = min(k, len(relevants))
        
        # Cắt mảng (Slicing) - Không cần tính lại từ đầu
        r_k = relevants[:actual_k]
        
        # 1. Hit Ratio@K
        # Kiểm tra xem có số 1 nào trong mảng r_k không
        metrics[f'Hit@{k}'] = 1 if np.any(r_k) else 0
        
        # 2. Precision@K
        # Dùng giá trị cuối cùng của mảng cumulative đã cắt (nhanh hơn sum lại)
        metrics[f'P@{k}'] = cumulative_relevant[actual_k - 1] / k if actual_k > 0 else 0
        
        # 3. NDCG@K
        dcg = np.sum(r_k / discounts[:actual_k])
        
        # IDCG (Cách tính của bạn: Sắp xếp lại r_k để 1 lên đầu)
        ideal_r_k = sorted(r_k, reverse=True)
        idcg = np.sum(ideal_r_k / discounts[:actual_k])
        
        metrics[f'NDCG@{k}'] = dcg / idcg if idcg > 0 else 0
        
        # 4. MRR@K
        # Tìm index đầu tiên bằng 1 trong phạm vi k
        # argmax trên mảng cắt r_k
        first_idx_k = np.argmax(r_k == 1)
        if r_k[first_idx_k] == 1:
            metrics[f'MRR@{k}'] = 1 / (first_idx_k + 1)
        else:
            metrics[f'MRR@{k}'] = 0
            
        # 5. MAP@K
        # MAP@K chỉ tính tổng precision trong top K chia cho tổng số Ground Truth
        if total_true_items > 0:
            # Lấy mảng precision đã tính sẵn, cắt đến k
            p_k = precisions_at_i[:actual_k]
            # Nhân với r_k để chỉ cộng những vị trí relevant
            metrics[f'MAP@{k}'] = np.sum(p_k * r_k) / total_true_items
        else:
            metrics[f'MAP@{k}'] = 0

    return pd.Series(metrics)
```
