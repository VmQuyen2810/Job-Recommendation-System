```python
import numpy as np
import gensim
from gensim.models import Doc2Vec
import os

# In ra phiên bản để double-check lần cuối
print(f"🛠️ Numpy version đang dùng: {np.__version__}")
print(f"🛠️ Gensim version đang dùng: {gensim.__version__}")
print("=" * 60)

# Danh sách 4 file model cần test (dựa trên ảnh bạn cung cấp)
doc2vec_models = [
    "word2vec_doc2vec_basic.model",
    "word2vec_doc2vec_basic_dbow.model",
    "word2vec_doc2vec_upgrade.model",
    "word2vec_doc2vec_upgrade_dbow.model"
]

for model_file in doc2vec_models:
    # Kiểm tra xem file có thực sự tồn tại ở thư mục hiện tại không
    if not os.path.exists(model_file):
        print(f"⚠️ BỎ QUA: Không tìm thấy file '{model_file}' tại thư mục hiện tại.")
        print("-" * 60)
        continue

    try:
        print(f"🔄 Đang tải: {model_file}...")
        
        # Thử load model
        model = Doc2Vec.load(model_file)
        
        # Nếu load qua dòng trên thành công, in ra một thông số nhỏ để xác nhận model chạy tốt
        print(f"✅ THÀNH CÔNG! Đã nạp '{model_file}' vào RAM.")
        print(f"   👉 Kích thước vector (Vector size): {model.vector_size}")
        print(f"   👉 Số lượng từ vựng (Vocab size): {len(model.wv.key_to_index)}")
        
    except Exception as e:
        print(f"❌ THẤT BẠI: Lỗi khi nạp '{model_file}'")
        print(f"   👉 Chi tiết lỗi: {str(e)}")
        
    print("-" * 60)
```
